```python
import pandas as pd

df = pd.read_csv("adult.data.csv")

df.head()
```

```python
race_count = df["race"].value_counts()

race_count
```

```python
average_age_men = df[df["sex"] == "Male"]["age"].mean()

average_age_men = round(average_age_men, 1)

average_age_men
```

```python
percentage_bachelors = (
    len(df[df["education"] == "Bachelors"]) / len(df) * 100
)

percentage_bachelors = round(percentage_bachelors, 1)

percentage_bachelors
```

```python
higher_education = df[
    df["education"].isin(["Bachelors", "Masters", "Doctorate"])
]

higher_education.head()
```

```python
lower_education = df[
    ~df["education"].isin(["Bachelors", "Masters", "Doctorate"])
]

lower_education.head()
```

```python
higher_education_rich = (
    len(higher_education[higher_education["salary"] == ">50K"])
    / len(higher_education)
    * 100
)

higher_education_rich = round(higher_education_rich, 1)

higher_education_rich
```

```python
lower_education_rich = (
    len(lower_education[lower_education["salary"] == ">50K"])
    / len(lower_education)
    * 100
)

lower_education_rich = round(lower_education_rich, 1)

lower_education_rich
```

```python
min_work_hours = df["hours-per-week"].min()

min_work_hours
```

```python
min_workers = df[
    df["hours-per-week"] == min_work_hours
]

min_workers.head()
```

```python
rich_percentage = (
    len(min_workers[min_workers["salary"] == ">50K"])
    / len(min_workers)
    * 100
)

rich_percentage = round(rich_percentage, 1)

rich_percentage
```

```python
country_count = df["native-country"].value_counts()

country_count
```

```python
rich_country_count = df[
    df["salary"] == ">50K"
]["native-country"].value_counts()

rich_country_count
```

```python
country_percentage = (
    rich_country_count / country_count * 100
)

country_percentage
```

```python
highest_earning_country_percentage = round(
    country_percentage.max(), 1
)

highest_earning_country_percentage
```

```python
india_rich = df[
    (df["native-country"] == "India") &
    (df["salary"] == ">50K")
]

india_rich.head()
```

```python
top_IN_occupation = india_rich["occupation"].value_counts().idxmax()

top_IN_occupation
```

```python
country_count = df["native-country"].value_counts()

rich_country_count = df[
    df["salary"] == ">50K"
]["native-country"].value_counts()

country_percentage = rich_country_count / country_count * 100

country_percentage
```

```python
highest_earning_country = country_percentage.idxmax()

highest_earning_country
```

```python
highest_earning_country_percentage = round(
    country_percentage.max(), 1
)

highest_earning_country_percentage
```

```python
india_rich = df[
    (df["native-country"] == "India") &
    (df["salary"] == ">50K")
]

top_IN_occupation = india_rich["occupation"].value_counts().idxmax()

top_IN_occupation
```

```python
print("Race count:")
print(race_count)

print("Average age men:", average_age_men)
print("Percentage Bachelors:", percentage_bachelors)
print("Higher education rich:", higher_education_rich)
print("Lower education rich:", lower_education_rich)
print("Minimum work hours:", min_work_hours)
print("Rich percentage:", rich_percentage)
print("Highest earning country:", highest_earning_country)
print("Highest earning country percentage:", highest_earning_country_percentage)
print("Top occupation in India:", top_IN_occupation)
```
